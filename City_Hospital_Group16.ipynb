City Hospital records every consultation in its information systems, but these records have not yet been used to understand how patients use the hospital's services. As the **Data Preprocessing (DP) Team**, our task is to prepare these data for the next analytics team, who will carry out descriptive analysis, patient segmentation and prediction.

The work is organised in three parts:

1. **Part A - Data Understanding & Cleaning:** assess the quality of the visit-level data and justify every cleaning decision.
2. **Part B - Feature Engineering & ABT Construction:** aggregate the visits into an Analytic Base Table (ABT) with one row per patient, documented in a data dictionary.
3. **Part C - Exploratory Visualization & Insights:** explore the ABT and the cleaned visits to answer management's questions.

No predictive models are built here; the goal is a clean, reproducible and well-documented starting point.

We check two kinds of duplicates: rows that are identical in every column, and visits of the same patient on the same day and in the same department.

```python
# Rows identical in all 16 columns
exact_dup = df[df.duplicated(keep=False)].sort_values(["Patient ID", "Visit Date"])
print("Exact duplicate rows (extra copies):", df.duplicated().sum())
exact_dup
```

```python
# Same patient, same day, same department (exact copies excluded)
key = ["Patient ID", "Visit Date", "Department"]
no_exact = df.drop_duplicates()
same_visit = no_exact[no_exact.duplicated(subset=key, keep=False)].sort_values(key)

print("Rows sharing patient, date and department:", len(same_visit))
same_visit[key + ["Consultation Duration", "Satisfaction Level", "Consultation Price", "Insurance Coverage"]].head(6)
```

* There are **8 exact duplicate rows**: each one is a perfect copy of another record, which can only be a data-entry or loading error.
* There are also **114 rows (57 pairs)** where a patient visited the same department on the same day. However, these rows have **different durations, ratings and prices**, so they are not copies of the same record but two separate consultations (for example, a follow-up on the same day).

<div style="background-color:#297681; padding:20px; border-radius:10px; border-left:8px solid #355473; color:#FFFFFF;">

**Removing the duplicated records:**
</div>

1. **Exact duplicates** are removed with `drop_duplicates()`, keeping the first occurrence.
2. **Same patient, date and department** rows are **kept**: removing them would delete real consultations and understate the patients' visits and costs.

```python
df = df.drop_duplicates().reset_index(drop=True)

print("Rows after removing exact duplicates:", df.shape[0])
print("Remaining exact duplicates:", df.duplicated().sum())
print("Unique patients:", df["Patient ID"].nunique())
```

All 500 patients are kept and the dataset now has **10,000 consultations**.

<div id="sec-2-3" style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #355473; color:#355473;">
  <b>2.3. Identify and Handle Missing Values</b>
</div>

```python
missing = pd.DataFrame({
    "Missing Values": df.isna().sum(),
    "Percentage (%)": (df.isna().mean() * 100).round(2)
})
missing[missing["Missing Values"] > 0].sort_values("Missing Values", ascending=False)
```

Six variables have missing values. Since every patient appears many times, we also check whether a value is missing in **all** of a patient's visits or only in **some** of them. This tells us whether the value can be recovered from the patient's other visits.

```python
miss_cols = ["Family History", "Insurance Provider", "Approximate Annual Income", "Age", "Insurance Coverage", "Education Level"]

# Share of missing visits per patient
patient_missing = df[miss_cols].isna().groupby(df["Patient ID"]).mean()

pd.DataFrame({
    "Patients missing in some visits": ((patient_missing > 0) & (patient_missing < 1)).sum(),
    "Patients missing in all visits": (patient_missing == 1).sum()
})
```

* `Age` and `Education Level` are missing only in some visits, so the value is known from the same patient's other visits.
* `Insurance Provider` is missing in some visits for 36 patients and in **all visits for 92 patients**.
* `Family History` is missing in **all visits for 99 patients** and never partially, so it is a patient-level gap, not a random loss of single values.
* `Approximate Annual Income` and `Insurance Coverage` are visit-level numeric values missing in a few visits.

To understand the 92 patients without any provider, we compare the share of the price paid by the insurer (`Insurance Coverage / Consultation Price`) for each provider:

```python
df["Coverage Ratio"] = (df["Insurance Coverage"] / df["Consultation Price"]).round(2)
df.groupby(df["Insurance Provider"].fillna("Missing"))["Coverage Ratio"].agg(["median", "min", "max"])
```

Each provider pays a **fixed share** of the price: Provider C 70%, A 80%, B 90% and D 100% (the few extreme values are errors treated in 2.4 and 2.5). For visits with no provider, the median share is **0%**, which means these patients **have no insurance**: the value is not lost, it does not apply.

```python
# Coverage ratio of the patients with no provider in any visit
no_provider_ids = patient_missing[patient_missing["Insurance Provider"] == 1].index
df[df["Patient ID"].isin(no_provider_ids)].groupby("Patient ID")["Coverage Ratio"].median().value_counts()
```

Of the 92 patients, **90 always pay the full price** (ratio 0). The other 2 have a constant ratio of 0.7 and 0.9, which identifies their provider as C and B.

<div style="background-color:#297681; padding:20px; border-radius:10px; border-left:8px solid #355473; color:#FFFFFF;">

**Handling the missing values:**
</div>

We impute instead of deleting rows: the gaps are small (except the two patient-level variables) and deleting visits would remove information needed to build the ABT. Each method uses the information already in the dataset:

* **Age and Education Level:** these attributes do not change during the six months, so a missing value is filled with the value recorded in the same patient's other visits.
* **Insurance Provider:** first filled from the same patient's other visits. For patients with no provider in any visit, the provider is inferred from the coverage ratio: ratio 0 becomes the new category **"No Insurance"**, as recommended for categorical variables where the value does not apply.
* **Family History:** missing for the whole patient, so it cannot be inferred. Following the class guidelines for categorical data, it is kept as a separate category, **"Not Reported"**, instead of being guessed or dropped.
* **Insurance Coverage:** calculated with the provider rule, `provider share × Consultation Price`, which holds in all the complete rows.
* **Approximate Annual Income:** the value changes from visit to visit, so it is filled with the **median income of the same patient**. The median is used because it is not affected by the extreme income values found in 2.5.

```python
# Static attributes: copy the value from the patient's other visits
df["Age"] = df["Age"].fillna(df.groupby("Patient ID")["Age"].transform("first"))
df["Education Level"] = df["Education Level"].fillna(df.groupby("Patient ID")["Education Level"].transform("first"))
df["Insurance Provider"] = df["Insurance Provider"].fillna(df.groupby("Patient ID")["Insurance Provider"].transform("first"))

# Patients with no provider at all: infer it from the share paid by the insurer
provider_rates = {"No Insurance": 0.0, "Provider C": 0.7, "Provider A": 0.8, "Provider B": 0.9, "Provider D": 1.0}
rate_to_provider = {0.0: "No Insurance", 0.7: "Provider C", 0.8: "Provider A", 0.9: "Provider B", 1.0: "Provider D"}
patient_ratio = df.groupby("Patient ID")["Coverage Ratio"].transform("median").round(1)
df["Insurance Provider"] = df["Insurance Provider"].fillna(patient_ratio.map(rate_to_provider))

# Family History: missing for the whole patient, kept as its own category
df["Family History"] = df["Family History"].fillna("Not Reported")

# Insurance Coverage: provider share x price
df["Insurance Coverage"] = df["Insurance Coverage"].fillna(df["Insurance Provider"].map(provider_rates) * df["Consultation Price"])

# Income: median of the same patient
df["Approximate Annual Income"] = df["Approximate Annual Income"].fillna(df.groupby("Patient ID")["Approximate Annual Income"].transform("median"))

df = df.drop(columns="Coverage Ratio")
print("Remaining missing values:", df.isna().sum().sum())
print()
print(df.drop_duplicates("Patient ID")["Insurance Provider"].value_counts())
```

No missing values remain. **90 patients (18%) are now identified as uninsured**, an important segment for the financial analysis.

  <b>2.4. Handling Inconsistent Data</b>
</div>

<div id="sec-2-4-1" style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #355473; color:#355473;">
  <b>2.4.1. Identifying Inconsistent Data</b>
</div>

**a) Category labels.** We list the categories of each text variable and check whether extra spaces or different capitalisation create duplicated labels.

```python
cat_cols = ["Gender", "City of Residence", "Profession", "Insurance Provider", "Family History", "Education Level", "Marital Status", "Department"]

pd.DataFrame({
    "Categories": df[cat_cols].apply(lambda s: sorted(s.unique())),
    "Unique labels": df[cat_cols].nunique(),
    "Unique after strip/lower": df[cat_cols].apply(lambda s: s.str.strip().str.lower().nunique())
})
```

All labels are written consistently: the number of categories does not change after removing spaces and capitalisation, so no label needs to be standardised.

**b) Patient attributes that change between visits.** Demographic attributes should be the same in every visit of a patient.

```python
static_cols = ["Age", "Gender", "City of Residence", "Profession", "Insurance Provider", "Family History", "Education Level", "Marital Status"]

changing = (df.groupby("Patient ID")[static_cols].nunique() > 1)
changing.sum()
```

```python
# Patients whose attributes change, with the number of visits for each value
changing_ids = changing[changing.any(axis=1)].index
df[df["Patient ID"].isin(changing_ids)].groupby(["Patient ID"])[["Age", "Gender", "Profession", "Marital Status"]].agg(lambda s: s.value_counts().to_dict())
```

12 patients have contradictory attributes: 5 in `Gender`, 4 in `Marital Status`, 2 in `Profession` and 1 in `Age`. In every case one value clearly dominates (e.g. patient 1307 is "Male" in 23 visits and "Female" in 1), so the odd values are isolated recording errors.

**c) Age and profession.** A child cannot be a lawyer, doctor, teacher or retired.

```python
minors_wrong = df[(df["Age"] < 18) & (df["Profession"] != "Student")]
minors_wrong.groupby("Patient ID")[["Age", "Profession", "Education Level", "Approximate Annual Income"]].agg(lambda s: s.unique().tolist())
```

Six children aged 1 to 11 are recorded as Lawyer, Doctor, Teacher or Retired. Their education level (Pre-school to Middle school) and their **income of £0** in every visit are consistent with their age, which shows that the profession is the wrong value. All other minors in the dataset are recorded as "Student".

**d) Satisfaction Level outside the 1-5 scale.**

```python
print("Visits with Satisfaction Level = 6:", (df["Satisfaction Level"] == 6).sum())
pd.crosstab(df["Department"], df["Satisfaction Level"])
```

**719 ratings are equal to 6**, which does not exist on the scale. They are not random: 715 of them come from **General Practice and ENT**, where 35% and 37% of the visits have this code, and these two departments have fewer ratings of 1-5 than the others. This suggests the 6 is a system code (for example "no answer") used by these departments rather than a real rating.

**e) Insurance Coverage greater than the price.** The insurer cannot pay more than the consultation costs.

```python
df[df["Insurance Coverage"] > df["Consultation Price"]][["Patient ID", "Department", "Insurance Provider", "Consultation Price", "Insurance Coverage"]]
```

In 6 visits the coverage is higher than the price. The prices are within the normal range of each department (see 2.5), while the coverages break the provider rule found in 2.3, so the **coverage** is the wrong value.

**f) Dates and negative values.**

```python
print("First visit:", df["Visit Date"].min().date(), "| Last visit:", df["Visit Date"].max().date())
print("Visits outside Jan-Jun 2024:", ((df["Visit Date"] < "2024-01-01") | (df["Visit Date"] > "2024-06-30")).sum())
print()
print("Negative values:")
print((df[num_cols] < 0).sum())
```

All visits fall inside the observation window and no variable has negative values.

<div id="sec-2-4-2" style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #355473; color:#355473;">
  <b>2.4.2. Correcting and Standardizing Inconsistent Data</b>
</div>

<div style="background-color:#297681; padding:20px; border-radius:10px; border-left:8px solid #355473; color:#FFFFFF;">

**Correcting the inconsistent values:**
</div>

* **Changing attributes:** replaced by the **most frequent value (mode) of the patient**, since the dominant value is the one recorded in most visits. `Age` is included; the remaining wrong age of patient 1198 (138) is treated with the other impossible ages in 2.5.
* **Minors with an adult profession:** set to **"Student"**, the profession of every other minor and consistent with their education and income.
* **Satisfaction Level = 6:** set to **missing** and **not imputed**. Changing it to 5 would make General Practice and ENT look more satisfied than they are, and imputing it would invent opinions that patients never gave. The satisfaction features in Part B are computed only on real ratings.
* **Coverage greater than price:** recalculated with the provider rule, `provider share × Consultation Price`.

```python
# Changing attributes: keep the patient's most frequent value
df["Age"] = df.groupby("Patient ID")["Age"].transform(lambda s: s.mode()[0])
df["Gender"] = df.groupby("Patient ID")["Gender"].transform(lambda s: s.mode()[0])
df["Profession"] = df.groupby("Patient ID")["Profession"].transform(lambda s: s.mode()[0])
df["Marital Status"] = df.groupby("Patient ID")["Marital Status"].transform(lambda s: s.mode()[0])

# Minors with an adult profession
df.loc[(df["Age"] < 18) & (df["Profession"] != "Student"), "Profession"] = "Student"

# Satisfaction outside the scale: missing, not imputed
df.loc[df["Satisfaction Level"] > 5, "Satisfaction Level"] = np.nan

# Coverage cannot exceed the price
wrong_coverage = df["Insurance Coverage"] > df["Consultation Price"]
df.loc[wrong_coverage, "Insurance Coverage"] = df.loc[wrong_coverage, "Insurance Provider"].map(provider_rates) * df.loc[wrong_coverage, "Consultation Price"]
```

Final check

```python
print("Patients with changing attributes:", (df.groupby("Patient ID")[static_cols].nunique() > 1).sum().sum())
print("Minors not recorded as Student:", ((df["Age"] < 18) & (df["Profession"] != "Student")).sum())
print("Ratings outside 1-5:", (df["Satisfaction Level"] > 5).sum(), "| unrated visits:", df["Satisfaction Level"].isna().sum())
print("Coverage greater than price:", (df["Insurance Coverage"] > df["Consultation Price"]).sum())
print("Minimum number of ratings per patient:", df.groupby("Patient ID")["Satisfaction Level"].count().min())
```

Every patient still has at least 9 valid ratings, so all patients get satisfaction features in the ABT.

<div id="sec-2-5" style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #355473; color:#355473;">
  <b>2.5. Identifying and Handling Outliers</b>
</div>

We detect outliers with the three methods seen in class: **boxplots**, the **IQR rule** (values below Q1 − 1.5×IQR or above Q3 + 1.5×IQR) and the **Z-score** (|z| > 3). `Satisfaction Level` is left out because it is now restricted to 1-5.

```python
out_cols = ["Age", "Consultation Duration", "Approximate Annual Income", "Consultation Price", "Insurance Coverage"]

df[out_cols].plot(kind="box", subplots=True, layout=(1, 5), figsize=(18, 4), color="#355473")
plt.suptitle("Boxplots of the numeric variables before outlier treatment")
plt.tight_layout()
plt.show()
```

```python
Q1 = df[out_cols].quantile(0.25)
Q3 = df[out_cols].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
z = (df[out_cols] - df[out_cols].mean()) / df[out_cols].std()

pd.DataFrame({
    "Lower bound": lower.round(1),
    "Upper bound": upper.round(1),
    "IQR outliers": ((df[out_cols] < lower) | (df[out_cols] > upper)).sum(),
    "Z-score outliers": (z.abs() > 3).sum(),
    "Max": df[out_cols].max().round(1)
})
```

The statistical rules flag values but do not tell us whether they are errors, so each variable is analysed with domain knowledge (as discussed in class, not every outlier is an error):

* **Age:** 36 visits above the upper bound of 153 years.
* **Consultation Duration:** 10 visits above 173 minutes, up to 600.
* **Approximate Annual Income:** the IQR flags 2,322 visits, but most are the £0 incomes of students and the incomes of a few high earners. Only 21 values (Z-score) are in the millions.
* **Consultation Price:** the IQR flags 240 visits because prices depend strongly on the department, so we inspect prices per department.
* **Insurance Coverage:** follows the price, so its outliers are explained by the price analysis.

**Age**

```python
df[df["Age"] > 110].groupby("Patient ID")[["Age", "Profession", "Education Level", "Marital Status", "Approximate Annual Income"]].first()
```

Five patients have impossible ages (118 to 195). All look like an extra "1" typed before the real age: patient 1198 is recorded with **38 years in one visit and 138 in the others**, and patient 1207 (118) is a single High-school student with no income, which fits an 18-year-old. Removing 100 years gives ages (18, 38, 42, 58 and 95) that are coherent with each patient's profession, education and marital status.

**Consultation Duration**

```python
print("Visits longer than 120 minutes:", (df["Consultation Duration"] > 120).sum())
print(df.loc[df["Consultation Duration"] > 120, ["Department", "Consultation Duration"]].sort_values("Consultation Duration").to_string())
print()
print("Duration of the remaining visits:")
print(df.loc[df["Consultation Duration"] <= 120, "Consultation Duration"].describe().round(1))
```

Every other consultation lasts between 15 and 120 minutes, in all departments. The 10 longer visits (228 to 600 minutes, i.e. up to 10 hours) are isolated and have no clear correction, so they are treated as recording errors.

**Approximate Annual Income**

```python
high_income = df[df["Approximate Annual Income"] > 60000]
print("Patients with income above £60,000:", high_income["Patient ID"].nunique())
print()

# Typical income of these patients compared with their extreme values
df[df["Patient ID"].isin(high_income["Patient ID"])].groupby("Patient ID")["Approximate Annual Income"].agg(
    visits="size",
    median="median",
    values_above_1M=lambda s: (s > 1000000).sum(),
    max="max")
```

Workers usually earn between £30,000 and £60,000. Ten patients (all professionals) earn £50,000-£120,000 consistently over all their visits: these are **real high earners**, a rare but valid group that is kept. However, **21 values are above £1 million** (about £1.15M and £11.5M). They belong to the same ten patients and are roughly **10 and 100 times** their usual income, a typical unit/typing error.

**Consultation Price**

```python
df.groupby("Department")["Consultation Price"].describe()[["min", "50%", "max"]].round(1)
```

```python
df[df["Consultation Price"] > 1000][["Patient ID", "Department", "Insurance Provider", "Consultation Price", "Insurance Coverage"]]
```

Each department has its own price range (e.g. General Practice and ENT £50-£100, Emergency £200-£400), so the global IQR flags normal Emergency prices. The real errors are **7 prices above £25,000**. For the two insured visits, the coverage still follows the provider rule for a price about **100 times smaller**: patient 1175 (Provider A, 80%) has a coverage of £319.34, which corresponds to a price of £399.18. The prices divided by 100 also fall inside the range of their department.

<div style="background-color:#297681; padding:20px; border-radius:10px; border-left:8px solid #355473; color:#FFFFFF;">

**To address each outlier, the following decisions were made:**
</div>

| Variable | Problem | Decision | Reason |
|---|---|---|---|
| `Age` | 5 patients aged 118-195 | **Correct**: subtract 100 | Extra leading "1"; confirmed by patient 1198 and by profession/education |
| `Consultation Duration` | 10 visits of 228-600 min | **Impute**: department median | No clear correction; median is robust and keeps the visit |
| `Approximate Annual Income` | 21 values above £1M | **Impute**: patient median | ×10 / ×100 errors; the patient's own median keeps them as high earners |
| `Approximate Annual Income` | £0 incomes and £60k-£120k incomes | **Keep** | Students/children and genuine high earners |
| `Consultation Price` | 7 prices above £25,000 | **Correct**: coverage ÷ provider share if insured, price ÷ 100 if uninsured | ×100 error confirmed by the coverage and the department range |
| `Insurance Coverage` | High values in expensive departments | **Keep** | Follow the provider rule on valid prices |

No visit or patient is removed in this step.

```python
# Age: remove the extra leading "1"
df.loc[df["Age"] > 110, "Age"] = df.loc[df["Age"] > 110, "Age"] - 100

# Duration: set the impossible values to missing and fill with the department median
df.loc[df["Consultation Duration"] > 120, "Consultation Duration"] = np.nan
df["Consultation Duration"] = df["Consultation Duration"].fillna(df.groupby("Department")["Consultation Duration"].transform("median"))

# Income: set the million values to missing and fill with the patient median
df.loc[df["Approximate Annual Income"] > 1000000, "Approximate Annual Income"] = np.nan
df["Approximate Annual Income"] = df["Approximate Annual Income"].fillna(df.groupby("Patient ID")["Approximate Annual Income"].transform("median"))

# Price: recover it from the coverage when insured, otherwise divide by 100
wrong_price = df["Consultation Price"] > 1000
insured = df["Insurance Provider"] != "No Insurance"
df.loc[wrong_price & insured, "Consultation Price"] = df.loc[wrong_price & insured, "Insurance Coverage"] / df.loc[wrong_price & insured, "Insurance Provider"].map(provider_rates)
df.loc[wrong_price & ~insured, "Consultation Price"] = df.loc[wrong_price & ~insured, "Consultation Price"] / 100
```

Final check

```python
df[out_cols].plot(kind="box", subplots=True, layout=(1, 5), figsize=(18, 4), color="#355473")
plt.suptitle("Boxplots of the numeric variables after outlier treatment")
plt.tight_layout()
plt.show()
```

```python
expected_coverage = df["Insurance Provider"].map(provider_rates) * df["Consultation Price"]
print("Visits breaking the coverage rule:", ((df["Insurance Coverage"] - expected_coverage).abs() > 0.01).sum())
print("Maximum age:", df["Age"].max(), "| maximum duration:", df["Consultation Duration"].max())
print("Maximum income: £{:,.0f} | maximum price: £{:,.2f}".format(df["Approximate Annual Income"].max(), df["Consultation Price"].max()))
print()
print("Final dataset:", df.shape, "| patients:", df["Patient ID"].nunique())
```

After cleaning, every value is plausible and the coverage rule holds in all 10,000 visits. The remaining points outside the whiskers (students' £0 income, high earners, expensive departments) are genuine and were kept on purpose.

Correcting the data types: `Age` only needed to be a float while it had missing values.

```python
df["Age"] = df["Age"].astype(int)
df["Consultation Duration"] = df["Consultation Duration"].astype(int)
df.dtypes
```

<div id="sec-3" style="background-color:#355473; padding:20px; border-radius:10px; border-left:8px solid #297681;">
  <h3 style="margin:0; color:#FFFFFF;"><b>3. Part B - Feature Engineering & ABT Construction</b></h3>
</div>

<div id="sec-3-1" style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #355473; color:#355473;">
  <b>3.1. Patient-Level ABT Construction</b>
</div>

The cleaned data has one row per visit. To build the ABT we **aggregate** the visits by `Patient ID`, so each patient becomes a single row. Before aggregating, we add three visit-level variables: the amount paid by the patient, the visit month and whether the rating was low (1 or 2).

The reference date for recency is **1 July 2024**, the day after the observation window ends.

```python
reference_date = pd.Timestamp("2024-07-01")

df["Out of Pocket"] = df["Consultation Price"] - df["Insurance Coverage"]
df["Visit Month"] = df["Visit Date"].dt.to_period("M")
df["Low Rating"] = (df["Satisfaction Level"] <= 2).astype(float).where(df["Satisfaction Level"].notna())

patients = df.groupby("Patient ID")
```

<div id="sec-3-1-1" style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #355473; color:#355473;">
  <b>3.1.1. Static Attributes</b>
</div>

The demographic and socio-economic attributes are now the same in every visit, so we take the first value. Income varies between visits, so we use the patient's **median** income. Two extra features are created: `age_group` (binning, Chapter 5) and `has_insurance`.

```python
static = patients.agg(
    age=("Age", "first"),
    gender=("Gender", "first"),
    city=("City of Residence", "first"),
    profession=("Profession", "first"),
    education_level=("Education Level", "first"),
    marital_status=("Marital Status", "first"),
    family_history=("Family History", "first"),
    insurance_provider=("Insurance Provider", "first"),
    annual_income=("Approximate Annual Income", "median"))

static["age_group"] = pd.cut(static["age"], bins=[-1, 17, 39, 64, 120], labels=["0-17", "18-39", "40-64", "65+"])
static["has_insurance"] = (static["insurance_provider"] != "No Insurance").astype(int)
static.head()
```

<div id="sec-3-1-2" style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #355473; color:#355473;">
  <b>3.1.2. Visit Behaviour Features</b>
</div>

```python
visits = patients.agg(
    n_visits=("Visit Date", "size"),
    n_departments=("Department", "nunique"),
    most_frequent_department=("Department", lambda s: s.value_counts().idxmax()),
    avg_duration=("Consultation Duration", "mean"),
    total_duration=("Consultation Duration", "sum"),
    n_emergency_visits=("Department", lambda s: (s == "Emergency").sum()))

visits["avg_duration"] = visits["avg_duration"].round(1)
visits.head()
```

<div id="sec-3-1-3" style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #355473; color:#355473;">
  <b>3.1.3. Financial Features</b>
</div>

```python
financial = patients.agg(
    total_price=("Consultation Price", "sum"),
    avg_price=("Consultation Price", "mean"),
    total_coverage=("Insurance Coverage", "sum"),
    total_out_of_pocket=("Out of Pocket", "sum"),
    avg_out_of_pocket=("Out of Pocket", "mean"))

# Share of the patient's total cost paid by the insurer
financial["coverage_share"] = financial["total_coverage"] / financial["total_price"]
financial = financial.round(2)
financial.head()
```

<div id="sec-3-1-4" style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #355473; color:#355473;">
  <b>3.1.4. Satisfaction Features</b>
</div>

Computed only on valid ratings (1-5); the removed "6" codes are counted separately in `n_unrated_visits`.

```python
satisfaction = patients.agg(
    avg_satisfaction=("Satisfaction Level", "mean"),
    min_satisfaction=("Satisfaction Level", "min"),
    share_low_ratings=("Low Rating", "mean"),
    n_rated_visits=("Satisfaction Level", "count"))

satisfaction["n_unrated_visits"] = patients.size() - satisfaction["n_rated_visits"]
satisfaction = satisfaction.round(2)
satisfaction.head()
```

<div id="sec-3-1-5" style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #355473; color:#355473;">
  <b>3.1.5. Recency and Activity Features</b>
</div>

`visits_per_month` divides the number of visits by the 6 months of the observation window. Every patient's first visit is in January or February, so all patients were observed during practically the whole period and the measure is comparable between them.

```python
recency = patients.agg(
    first_visit=("Visit Date", "min"),
    last_visit=("Visit Date", "max"),
    active_months=("Visit Month", "nunique"))

recency["days_since_first_visit"] = (reference_date - recency["first_visit"]).dt.days
recency["days_since_last_visit"] = (reference_date - recency["last_visit"]).dt.days
recency["visits_per_month"] = (visits["n_visits"] / 6).round(2)
recency = recency.drop(columns=["first_visit", "last_visit"])
recency.head()
```

Joining all the feature groups into the ABT:

```python
abt = static.join([visits, financial, satisfaction, recency]).reset_index()
abt = abt.rename(columns={"Patient ID": "patient_id"})

print("ABT shape:", abt.shape)
print("Unique patients:", abt["patient_id"].nunique(), "| missing values:", abt.isna().sum().sum())
abt.head()
```

The ABT has **exactly one row per patient (500 rows)**, 33 columns and no missing values.

Satisfaction is **almost identical across segments** (2.94-3.05). Being uninsured does not lower satisfaction. Since no segment stands out, dissatisfaction seems to be a **hospital-wide issue** rather than a problem of a particular group.

  <b>4.5. Management Questions and Key Insights</b>

**1. Which departments are the busiest, and how does demand change from January to June?**
Psychiatry and General Practice are the busiest, each with about 1,350 consultations (twice any other department). Demand is stable over the six months at 53-57 consultations per day; the dip in February is explained by the shorter month.

**2. Which patient groups use which departments most?**
All groups use the departments in a very similar way, led by Psychiatry and General Practice. Small differences exist: children use General Practice and Gastroenterology more, older patients use Pulmonology more, and men and "Other" use Neurology and Endocrinology a little more than women.

**3. Which departments and patient groups report the lowest satisfaction?**
General Practice (2.85), Pulmonology and Neurology (2.94) have the lowest averages, and patients with a family history of hypertension or heart disease are the least satisfied segments. However, differences are small and satisfaction is low across the whole hospital (average 2.99, ~40% low ratings), so it should be treated as a general priority. General Practice and ENT also need a better rating collection process, since about 35% of their ratings were invalid.

**4. How much of the consultation cost is covered by insurance, and how much do patients pay themselves?**
Insurers cover 70% of the £1.65M total and patients pay the other 30% (£0.50M). The 90 uninsured patients (18%) carry 60% of the out-of-pocket costs.

* **Segmentation ideas:** combine age group, insurance status, visit intensity (`n_visits`, `n_departments`) and cost (`total_price`, `coverage_share`), e.g. *uninsured frequent users*, *older multi-department patients* and *young low-use patients*.
