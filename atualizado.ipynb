  <b>2.3. Identify and Handle Missing Values </b>
</div>

Before addressing missing data, we run `df.info()` to inspect the general schema, data types, and total volume across all 16 features. Because this high-level view only displays the count of non-null records, we follow it with a dedicated missingness summary to quantify both the exact number of missing entries and their relative proportions, identifying which columns require targeted imputation.

```python
df.info()
```

```python
missing_count = df.isnull().sum()
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    'Missing Values': missing_count,
    'Percentage (%)': missing_percent.round(2)
})

missing_summary[missing_summary['Missing Values'] > 0]
```

The summary reveals that missing values are concentrated across six features, clearly splitting into two patterns: `Family History` (19.94%) and `Insurance Provider` (19.08%) exhibit substantial missingness, which in a healthcare domain often reflects structural absence—such as having no relevant genetic history or lacking private insurance—making them candidates for dedicated categorical markers rather than raw deletion, whereas `Approximate Annual Income` (1.53%), `Age` (0.56%), `Insurance Coverage` (0.49%), and `Education Level` (0.29%) show negligible gaps that can be safely addressed via standard statistical or grouped imputation without distorting the underlying distributions.

```python
plt.figure(figsize=(10, 4))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis', yticklabels=False)
plt.title('Missing Values Matrix (Yellow = Missing)')
plt.show()
```

The missingness matrix reveals that data gaps are largely confined to Insurance Provider and Family History, displaying contrasting structures: Insurance Provider exhibits a continuous vertical block indicative of systematic missingness for specific patient profiles, while Family History shows a fragmented, row-wise distribution consistent with random non-reporting during consultations. Demographic and financial variables (Age, Approximate Annual Income, Insurance Coverage, and Education Level) present only marginal, isolated missing entries.


<div style="background-color:#297681; padding:20px; border-radius:10px; border-left:8px"> 
    
**Handling the missing values:**

We use imputation instead of deleting rows because the missing values are few (between 0.3% and 1.5% of the visits) and we need all 500 patients and all their visits to build the ABT. Removing rows would throw away useful information, such as a patient's other visits.

The missing values can also be filled reliably with information that is already in the dataset: age, education and insurance provider do not change for a patient, the insurance coverage follows a fixed rule (each provider pays a fixed share of the price), and income can be replaced by the patient's median.

```python
df_imputed = df.copy() 

# Age, education and provider do not change for a patient,
# so a missing value is filled with the value from another visit of the same patient
for col in ['Age', 'Education Level', 'Insurance Provider']:
    df_imputed[col] = df_imputed.groupby('Patient ID')[col].transform(lambda s: s.ffill().bfill())

# A provider always pays the same share of the price (e.g. Provider A pays 80% in every visit),
# so coverage / price tells us which provider it is
rate_by_provider = {0.0: 'None', 0.8: 'Provider A', 0.9: 'Provider B', 0.7: 'Provider C', 1.0: 'Provider D'}
ratio = (df_imputed['Insurance Coverage'] / df_imputed['Consultation Price']).round(1)

# Patients with no provider in any visit: use the share to find it
df_imputed['Insurance Provider'] = df_imputed['Insurance Provider'].fillna(ratio.map(rate_by_provider))

# The same rule the other way round: the share paid by each provider
provider_rates = {'None': 0.0, 'Provider A': 0.8, 'Provider B': 0.9, 'Provider C': 0.7, 'Provider D': 1.0}
expected_rate = df_imputed['Insurance Provider'].map(provider_rates)

# Missing coverage = provider share x consultation price
df_imputed['Insurance Coverage'] = df_imputed['Insurance Coverage'].fillna(expected_rate * df_imputed['Consultation Price'])

# Income changes between visits of the same patient, so we use the patient's median
income_median = df_imputed.groupby('Patient ID')['Approximate Annual Income'].transform('median')
df_imputed['Approximate Annual Income'] = df_imputed['Approximate Annual Income'].fillna(income_median)

# Family history cannot be worked out from other columns, so it is marked as Unknown
df_imputed['Family History'] = df_imputed['Family History'].fillna('Unknown')

# Check: every column should show 0 missing values
print("Missing values after imputation:")
print(df_imputed.isnull().sum())
```

To handle the missing values we used imputation, based on the patient's other visits and on rules we found in the data. Group-mean imputation was rejected because it would hide real differences between patients.

* **Age and Education Level:**
  These do not change for a patient during the six months of data, so a missing value is copied from another visit of the same patient (`ffill().bfill()` grouped by `Patient ID`). This restores the real value instead of inventing a new one.

* **Insurance Provider:**
  A missing provider is first filled with the provider of the same patient's other visits. For patients with no provider in any visit, we infer it from the share of the price that the insurance paid (`Insurance Coverage / Consultation Price`). Each provider always pays the same share (None 0%, C 70%, A 80%, B 90%, D 100%), so this share identifies the provider. The ratio table in 1.5 shows the evidence: the median ratio of each provider equals its rate, and only 8 rows break the rule.

* **Insurance Coverage:**
  A missing coverage is calculated as the provider's share times the consultation price. Since this rule holds in all the complete rows, the result is exact. It is applied in 1.8 after the price has been corrected, because a wrong price would give a wrong coverage.

* **Approximate Annual Income:**
  Income changes slightly from visit to visit, so a missing value is replaced by the median income of the same patient (`transform('median')`). The median is not affected by a single unusual value, every patient has at least one value, and patients with zero income (children and students) keep their zero. The same median is used as the income feature in the ABT (2.1.1).

* **Family History:**
  This cannot be inferred from any other column, so a missing value is labelled `'Unknown'` instead of being guessed. The value `'None'` is not missing: it means the patient has no family disease and is kept as a category.

After resolving all inconsistencies during the initial phase of the project, the variables **age, gender, city of residence, profession, insurance provider, family history, education level and marital status** were classified as static attributes. Since these variables remain unchanged across all records belonging to the same patient, we can simply select the first recorded value for each attribute when consolidating the dataset into one row per patient.

```python
ref_date = pd.Timestamp('2024-07-01')

# Low rating = 1 or 2 (missing ratings stay missing)
df['Low Rating'] = (df['Satisfaction Level'] <= 2).astype(float).where(df['Satisfaction Level'].notna())

# Invalid rating = rating of 6 in the raw data (in df they are already missing)
df['Invalid Rating'] = (data.loc[df.index, 'Satisfaction Level'] > 5).astype(int)
```

<div style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #355473; color:#355473;">

</div>

```python
static_attributes = df.groupby("Patient ID").agg({
    "Age": "first",
    "Gender": "first",
    "City of Residence": "first",
    "Profession": "first",
    "Education Level": "first",
    "Marital Status": "first",
    "Family History": "first",
    "Approximate Annual Income": "median",
    "Insurance Provider": "first"
}).reset_index()
static_attributes.columns = ['Patient_ID', 'Age', 'Gender', 'City', 'Profession', 'Education_Level', 'Marital_Status', 'Family_History', 'Annual_Income', 'Insurance_Provider']

static_attributes.head()
```

<div style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #355473; color:#355473;">
  <b>3.1.2 Visit Behaviour Features </b>
</div>


These features describe how each patient uses the hospital: how many consultations they had, in how many different departments, which department they visit most and how long their consultations last.

```python
visits = df.groupby('Patient ID').agg({
    'Visit Date': 'count',
    'Department': ['nunique', lambda s: s.mode()[0]],
    'Consultation Duration': ['mean', 'sum']
}).reset_index()
visits.columns = ['Patient_ID', 'N_Visits', 'N_Departments', 'Most_Visited_Department', 'Avg_Duration', 'Total_Duration']
visits.head()
```

<div style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #355473; color:#355473;">
  <b>3.1.3 Financial Features </b>
</div>


The financial features show how much each patient's consultations cost, how much the insurance paid and how much the patient paid (out-of-pocket = price - coverage). The `coverage_share` is the part of the total cost paid by the insurance.

```python
# Out-of-pocket amount of each visit (price - coverage)
df['Out of Pocket'] = df['Consultation Price'] - df['Insurance Coverage']

financial = df.groupby('Patient ID').agg({
    'Consultation Price': ['sum', 'mean'],
    'Insurance Coverage': 'sum',
    'Out of Pocket': ['sum', 'mean']
}).reset_index()
financial.columns = ['Patient_ID', 'Total_Price', 'Avg_Price', 'Total_Coverage', 'Total_Out_Of_Pocket', 'Avg_Out_Of_Pocket']

financial['Coverage_Share'] = financial['Total_Coverage'] / financial['Total_Price']
financial.head()
```

<div style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #355473; color:#355473;">
  <b>3.1.4 Satisfaction Features </b>


The satisfaction features use only the valid ratings (1-5); the missing ones are ignored by `mean` and `min`. A low rating is a rating of 1 or 2. We also keep the number of invalid ratings of each patient as a data quality flag.

```python
satisfaction = df.groupby('Patient ID').agg({
    'Satisfaction Level': ['mean', 'median', 'min', 'max'],
    'Low Rating': 'mean',
    'Invalid Rating': 'sum'
}).reset_index()
satisfaction.columns = ['Patient_ID', 'Avg_Satisfaction', 'Median_Satisfaction', 'Min_Satisfaction', 'Max_Satisfaction', 'Share_Low_Ratings', 'N_Invalid_Ratings']
satisfaction.head()
```

<div style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #355473; color:#355473;">
  <b>3.1.5 Recency and Activity Featuress </b>


Recency is measured in days between the first and the last visit and the reference date (1 July 2024). Activity is measured by the number of different months with at least one visit and by the average number of visits per month in the six months of data.

```python
recency = df.groupby('Patient ID').agg({
    'Visit Date': ['min', 'max', lambda s: s.dt.month.nunique()]
}).reset_index()
recency.columns = ['Patient_ID', 'First_Visit', 'Last_Visit', 'N_Active_Months']

# Number of visits of each patient
n_visits = recency['Patient_ID'].map(df.groupby('Patient ID').size())

# Recency: days between the first/last visit and the reference date (1 July 2024)
recency['Recency_Days'] = (ref_date - recency['Last_Visit']).dt.days
recency['Days_Since_First_Visit'] = (ref_date - recency['First_Visit']).dt.days

# Tenure: days between the first and the last visit
recency['Tenure_Days'] = (recency['Last_Visit'] - recency['First_Visit']).dt.days

# Average interval between consecutive visits
recency['Avg_Days_Between_Visits'] = (recency['Tenure_Days'] / (n_visits - 1)).round(2)

# Activity: visits per month in the six months of data
recency['Visits_Per_Month'] = (n_visits / 6).round(2)
recency.head()
```

**Joining the five groups into the ABT**

```python
abt = static_attributes.merge(visits, on='Patient_ID').merge(financial, on='Patient_ID').merge(satisfaction, on='Patient_ID').merge(recency, on='Patient_ID')
abt = abt.round(2)

print("ABT shape:", abt.shape)
print("One row per patient:", abt['Patient_ID'].is_unique)
print("Missing values:", abt.isna().sum().sum())
```
