We began by identifying and removing duplicate entries, ensuring that repeated observations would not compromise the accuracy of the missing-value treatment performed afterwards.

Following the initial inspection with `df.info()`, we applied `df.map(type).nunique()` to quickly examine the data types present in each column and check for potential inconsistencies in the dataset's structure before moving on to duplicate detection.


```python
df.map(type).nunique() 
#this shows us where in the dataset the values are not unique
```

```python
# Check for fully identical rows to avoid counting the exact same record twice
exact_duplicates = df[df.duplicated()]
print(f"Exact duplicate rows: {exact_duplicates.shape[0]}")
display(exact_duplicates)

# Check for duplicate visits to avoid counting the same patient multiple times in the same day and department
visit_duplicates = df[df.duplicated(subset=["Patient ID", "Visit Date", "Department"], keep=False)]
print(f"\nDuplicate visits (same patient, date, department): {visit_duplicates.shape[0]}")
display(visit_duplicates.head(10))
```

The initial inspection revealed **8 exact duplicate rows** (fully identical across all features), as well as **130 records** involved in duplicate visits (same `Patient ID`, `Visit Date`, and `Department`). Therefore, we need to handle these redundant entries.



<div style="background-color:#297681; padding:20px; border-radius:10px; border-left:8px"> 
    
**Cleaning the duplicated records:**

To clean the data:
1. **Exact Duplicates:** We remove the 8 redundant copies using `df.drop_duplicates()` to eliminate pure data ingestion errors while keeping the first occurrence.
2. **Duplicate Visits:** We then apply `df.drop_duplicates(subset=['Patient ID', 'Visit Date', 'Department'], keep='first')` to keep only the initial entry and remove redundant visits for the same patient on the same day and department, preserving cohort consistency without losing unique patient profiles.

```python
df = df.drop_duplicates().copy()

print(f"Rows after removing exact duplicates: {len(df):,} | patients: {df['Patient ID'].nunique()}")
print(f"Fewest visits left for any patient: {df.groupby('Patient ID').size().min()}")
```

We use `df.drop_duplicates()` with a specific subset to remove multiple visit entries for the same patient in the same department on the same date. The `keep='first'` argument keeps only the first occurrence and removes the rest.

```python
df = df.drop_duplicates(subset=["Patient ID", "Visit Date", "Department"], keep='first')
print(f"Final shape after removing duplicate visits: {df.shape[0]:,} rows and {df.shape[1]} features.")
```

We removed identical records across all fields by keeping only the first occurrence, as they represent redundant data-entry errors. This de-duplication step preserved the entire cohort without losing any patients, ensuring that every individual maintained at least the minimum required number of recorded visits

  <b>2.3. Identify and Handle Missing Values </b>
</div>

Before addressing missing data, we run `df.info()` to inspect the general schema, data types, and total volume across all 16 features. Because this high-level view only displays the count of non-null records, we follow it with a dedicated missingness summary to quantify both the exact number of missing entries and their relative proportions, identifying which columns require targeted imputation.

```python
df.info()
```

```python
missing_count = df.isnull().sum()
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    'Missing Values': missing_count,
    'Percentage (%)': missing_percent.round(2)
})

missing_summary[missing_summary['Missing Values'] > 0]
```

The summary reveals that missing values are concentrated across six features, clearly splitting into two patterns: `Family History` (19.94%) and `Insurance Provider` (19.08%) exhibit substantial missingness, which in a healthcare domain often reflects structural absence—such as having no relevant genetic history or lacking private insurance—making them candidates for dedicated categorical markers rather than raw deletion, whereas `Approximate Annual Income` (1.53%), `Age` (0.56%), `Insurance Coverage` (0.49%), and `Education Level` (0.29%) show negligible gaps that can be safely addressed via standard statistical or grouped imputation without distorting the underlying distributions.

```python
plt.figure(figsize=(10, 4))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis', yticklabels=False)
plt.title('Missing Values Matrix (Yellow = Missing)')
plt.show()
```

The missingness matrix reveals that data gaps are largely confined to Insurance Provider and Family History, displaying contrasting structures: Insurance Provider exhibits a continuous vertical block indicative of systematic missingness for specific patient profiles, while Family History shows a fragmented, row-wise distribution consistent with random non-reporting during consultations. Demographic and financial variables (Age, Approximate Annual Income, Insurance Coverage, and Education Level) present only marginal, isolated missing entries.


<div style="background-color:#297681; padding:20px; border-radius:10px; border-left:8px"> 
    
**Handling the missing values:**

We use imputation instead of deleting rows because the missing values are few (between 0.3% and 1.5% of the visits) and we need all 500 patients and all their visits to build the ABT. Removing rows would throw away useful information, such as a patient's other visits.

The missing values can also be filled reliably with information that is already in the dataset: age, education and insurance provider do not change for a patient, the insurance coverage follows a fixed rule (each provider pays a fixed share of the price), and income can be replaced by the patient's median.

```python
df_imputed = df.copy() 

# Age, education and provider do not change for a patient,
# so a missing value is filled with the value from another visit of the same patient
for col in ['Age', 'Education Level', 'Insurance Provider']:
    df_imputed[col] = df_imputed.groupby('Patient ID')[col].transform(lambda s: s.ffill().bfill())

# A provider always pays the same share of the price (e.g. Provider A pays 80% in every visit),
# so coverage / price tells us which provider it is
rate_by_provider = {0.0: 'None', 0.8: 'Provider A', 0.9: 'Provider B', 0.7: 'Provider C', 1.0: 'Provider D'}
ratio = (df_imputed['Insurance Coverage'] / df_imputed['Consultation Price']).round(1)

# Patients with no provider in any visit: use the share to find it
df_imputed['Insurance Provider'] = df_imputed['Insurance Provider'].fillna(ratio.map(rate_by_provider))

# The same rule the other way round: the share paid by each provider
provider_rates = {'None': 0.0, 'Provider A': 0.8, 'Provider B': 0.9, 'Provider C': 0.7, 'Provider D': 1.0}
expected_rate = df_imputed['Insurance Provider'].map(provider_rates)

# Missing coverage = provider share x consultation price
df_imputed['Insurance Coverage'] = df_imputed['Insurance Coverage'].fillna(expected_rate * df_imputed['Consultation Price'])

# Income changes between visits of the same patient, so we use the patient's median
income_median = df_imputed.groupby('Patient ID')['Approximate Annual Income'].transform('median')
df_imputed['Approximate Annual Income'] = df_imputed['Approximate Annual Income'].fillna(income_median)

# Family history cannot be worked out from other columns, so it is marked as Unknown
df_imputed['Family History'] = df_imputed['Family History'].fillna('Unknown')

# Check: every column should show 0 missing values
print("Missing values after imputation:")
print(df_imputed.isnull().sum())
```

To handle the missing values we used imputation, based on the patient's other visits and on rules we found in the data. Group-mean imputation was rejected because it would hide real differences between patients.

* **Age and Education Level:**
  These do not change for a patient during the six months of data, so a missing value is copied from another visit of the same patient (`ffill().bfill()` grouped by `Patient ID`). This restores the real value instead of inventing a new one.

* **Insurance Provider:**
  A missing provider is first filled with the provider of the same patient's other visits. For patients with no provider in any visit, we infer it from the share of the price that the insurance paid (`Insurance Coverage / Consultation Price`). Each provider always pays the same share (None 0%, C 70%, A 80%, B 90%, D 100%), so this share identifies the provider. The ratio table in 1.5 shows the evidence: the median ratio of each provider equals its rate, and only 8 rows break the rule.

* **Insurance Coverage:**
  A missing coverage is calculated as the provider's share times the consultation price. Since this rule holds in all the complete rows, the result is exact. It is applied in 1.8 after the price has been corrected, because a wrong price would give a wrong coverage.

* **Approximate Annual Income:**
  Income changes slightly from visit to visit, so a missing value is replaced by the median income of the same patient (`transform('median')`). The median is not affected by a single unusual value, every patient has at least one value, and patients with zero income (children and students) keep their zero. The same median is used as the income feature in the ABT (2.1.1).

* **Family History:**
  This cannot be inferred from any other column, so a missing value is labelled `'Unknown'` instead of being guessed. The value `'None'` is not missing: it means the patient has no family disease and is kept as a category.

## <font color='turquoise'>1.5.1 Detecting Outliers in Numerical Variables

```python
# Select numerical columns from the DataFrame
num_cols = ["Age", "Consultation Duration", "Satisfaction Level", "Approximate Annual Income", "Consultation Price", "Insurance Coverage"]

# min and max already show some strange values (age 195, rating 6, price 39999...)
display(df[num_cols].describe().round(1))

# IQR rule (1.5 x IQR) for each variable
print()
for col in num_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    n_outliers = ((df[col] < q1 - 1.5 * iqr) | (df[col] > q3 + 1.5 * iqr)).sum()
    print(col, "| rows outside the IQR limits:", n_outliers)

df[num_cols].plot(kind="box", subplots=True, layout=(2, 3), figsize=(14, 8))
plt.suptitle("Boxplots of the numerical variables (before treating outliers)")
plt.tight_layout()
plt.show()
```

The describe table and the boxplots show a few values that are far away from the rest: ages up to 195, prices up to £39,999, incomes up to £12M, durations up to 600 minutes and a rating of 6.

The IQR rule alone is not good to decide what to change. It marks about 2,300 incomes and 241 prices as outliers, but most of them are normal values: many children and students have an income of 0, which makes the distribution very skewed, and the price depends on the department (see below). So we use the IQR as a first look and then check each variable with what is possible in the real world and with the variable description in the project statement.

## <font color='turquoise'> 1.5.2 Deciding How to Handle Outliers

| Variable | Outlier | Decision | Why |
|---|---|---|---|
| Age | 103 rows (5 patients) above 110 | Correct: subtract 100 | Typing error with an extra "1", confirmed by patient 1198 and by the profession and education |
| Consultation Price | 7 rows above £1000 | Correct: divide by 100 | The result is inside the normal range of the department |
| Approximate Annual Income | 21 rows above £1M | Correct: divide by 10 (£1M-£10M) or by 100 (above £10M) | The result is close to the highest normal incomes and to the patient's other visits |
| Income = 0 | 110 patients | Keep | Valid value (mostly children and students) |
| Consultation Duration | 10 rows above 120 min | Impute with the median of the department | The extreme values are not a clear multiple of a normal duration, so the real value is unknown |
| Satisfaction Level | 719 rows with a 6 | Set to missing | The rating is impossible and the real opinion is unknown. We do not change it to 5 because it would invent good ratings |
| Insurance Coverage | 6 rows higher than the price | Recompute: provider rate x price | All the other rows follow this rule exactly |

We do not remove any row or patient: the errors have a clear pattern or can be replaced, and removing visits would lose real information (and patients).

```python
# Age: extra "1" in front -> subtract 100
df.loc[df["Age"] > 110, "Age"] = df.loc[df["Age"] > 110, "Age"] - 100

# Price: 100 times too high -> divide by 100
df.loc[df["Consultation Price"] > 1000, "Consultation Price"] = df.loc[df["Consultation Price"] > 1000, "Consultation Price"] / 100

# Income: 100 times too high (above 10 million) or 10 times too high (above 1 million)
df.loc[df["Approximate Annual Income"] >= 10_000_000, "Approximate Annual Income"] = df.loc[df["Approximate Annual Income"] >= 10_000_000, "Approximate Annual Income"] / 100
df.loc[df["Approximate Annual Income"] >= 1_000_000, "Approximate Annual Income"] = df.loc[df["Approximate Annual Income"] >= 1_000_000, "Approximate Annual Income"] / 10

# Duration: above 120 -> missing -> median of the department
df.loc[df["Consultation Duration"] > 120, "Consultation Duration"] = np.nan
df["Consultation Duration"] = df["Consultation Duration"].fillna(df.groupby("Department")["Consultation Duration"].transform("median"))

# Satisfaction: 6 is not a valid rating -> missing
df.loc[df["Satisfaction Level"] == 6, "Satisfaction Level"] = np.nan

# Coverage: higher than the price -> provider rate x price
rate_by_provider = {"None": 0.0, "Provider A": 0.8, "Provider B": 0.9, "Provider C": 0.7, "Provider D": 1.0}
too_high = df["Insurance Coverage"] > df["Consultation Price"]
df.loc[too_high, "Insurance Coverage"] = df.loc[too_high, "Insurance Provider"].map(rate_by_provider) * df.loc[too_high, "Consultation Price"]

# check the result
display(df[num_cols].describe().round(1))
print("\nAges above 110:", (df["Age"] > 110).sum())
print("Prices above 1000:", (df["Consultation Price"] > 1000).sum())
print("Incomes above 1 million:", (df["Approximate Annual Income"] >= 1_000_000).sum())
print("Durations above 120:", (df["Consultation Duration"] > 120).sum())
print("Coverage higher than the price:", (df["Insurance Coverage"] > df["Consultation Price"]).sum())
print("Missing ratings (the old 6s):", df["Satisfaction Level"].isna().sum())
print("Rows and patients:", len(df), "|", df["Patient ID"].nunique())

df[num_cols].plot(kind="box", subplots=True, layout=(2, 3), figsize=(14, 8))
plt.suptitle("Boxplots of the numerical variables (after treating outliers)")
plt.tight_layout()
plt.show()
```

After the corrections all the values are inside the possible ranges and we still have the 10,008 visits of the 500 patients. The boxplots now show compact distributions without the extreme points.

Note for the ABT: the 719 ratings that became missing are ignored by `mean()` and `min()`. The share of low ratings should be calculated only with the valid ratings (for example with `dropna()` first).
