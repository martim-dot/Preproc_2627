  <b>2.3. Identify and Handle Missing Values </b>
</div>

Before addressing missing data, we run `df.info()` to inspect the general schema, data types, and total volume across all 16 features. Because this high-level view only displays the count of non-null records, we follow it with a dedicated missingness summary to quantify both the exact number of missing entries and their relative proportions, identifying which columns require targeted imputation.

```python
df.info()
```

```python
missing_count = df.isnull().sum()
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    'Missing Values': missing_count,
    'Percentage (%)': missing_percent.round(2)
})

missing_summary[missing_summary['Missing Values'] > 0]
```

The summary reveals that missing values are concentrated across six features, clearly splitting into two patterns: `Family History` (19.94%) and `Insurance Provider` (19.08%) exhibit substantial missingness, which in a healthcare domain often reflects structural absence, such as having no relevant genetic history or lacking private insurance, making them candidates for dedicated categorical markers rather than raw deletion, whereas `Approximate Annual Income` (1.53%), `Age` (0.56%), `Insurance Coverage` (0.49%), and `Education Level` (0.29%) show negligible gaps that can be safely addressed via standard statistical or grouped imputation without distorting the underlying distributions.

```python
plt.figure(figsize=(10, 4))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis', yticklabels=False)
plt.title('Missing Values Matrix (Yellow = Missing)')
plt.show()
```

The missingness matrix reveals that data gaps are largely confined to Insurance Provider and Family History, displaying contrasting structures: Insurance Provider exhibits a continuous vertical block indicative of systematic missingness for specific patient profiles, while Family History shows a fragmented, row-wise distribution consistent with random non-reporting during consultations. Demographic and financial variables (Age, Approximate Annual Income, Insurance Coverage, and Education Level) present only marginal, isolated missing entries.


```python
miss_cols = ["Family History", "Insurance Provider", "Approximate Annual Income", "Age", "Insurance Coverage", "Education Level"]

# Share of missing visits per patient
patient_missing = df[miss_cols].isna().groupby(df["Patient ID"]).mean()

pd.DataFrame({
    "Patients missing in some visits": ((patient_missing > 0) & (patient_missing < 1)).sum(),
    "Patients missing in all visits": (patient_missing == 1).sum()
})
```

* `Age` and `Education Level` are missing only in some visits, so the value is known from the same patient's other visits.
* `Insurance Provider` is missing in some visits for 36 patients and in **all visits for 92 patients**.
* `Family History` is missing in **all visits for 99 patients** and never partially, so it is a patient-level gap, not a random loss of single values.
* `Approximate Annual Income` and `Insurance Coverage` are visit-level numeric values missing in a few visits.

To understand the 92 patients without any provider, we compare the share of the price paid by the insurer (`Insurance Coverage / Consultation Price`) for each provider:

```python
df["Coverage Ratio"] = (df["Insurance Coverage"] / df["Consultation Price"]).round(2)
df.groupby(df["Insurance Provider"].fillna("Missing"))["Coverage Ratio"].agg(["median", "min", "max"])
```

<div style="background-color:#297681; padding:20px; border-radius:10px; border-left:8px"> 
    
**Handling the missing values:**

We use imputation instead of deleting rows because the missing values are few (between 0.3% and 1.5% of the visits) and we need all 500 patients and all their visits to build the ABT. Removing rows would throw away useful information, such as a patient's other visits.

The missing values can also be filled reliably with information that is already in the dataset: age, education and insurance provider do not change for a patient, the insurance coverage follows a fixed rule (each provider pays a fixed share of the price), and income can be replaced by the patient's median.

```python
df.head()
```

```python
df_imputed = df.copy() 

# Age, education and provider do not change for a patient,
# so a missing value is filled with the value from another visit of the same patient
for col in ['Age', 'Education Level', 'Insurance Provider']:
    df_imputed[col] = df_imputed.groupby('Patient ID')[col].transform(lambda s: s.ffill().bfill())

# A provider always pays the same share of the price (e.g. Provider A pays 80% in every visit),
# so coverage / price tells us which provider it is
rate_by_provider = {0.0: 'None', 0.8: 'Provider A', 0.9: 'Provider B', 0.7: 'Provider C', 1.0: 'Provider D'}
ratio = (df_imputed['Insurance Coverage'] / df_imputed['Consultation Price']).round(1)

# Patients with no provider in any visit: use the share to find it
df_imputed['Insurance Provider'] = df_imputed['Insurance Provider'].fillna(ratio.map(rate_by_provider))

# The same rule the other way round: the share paid by each provider
provider_rates = {'None': 0.0, 'Provider A': 0.8, 'Provider B': 0.9, 'Provider C': 0.7, 'Provider D': 1.0}
expected_rate = df_imputed['Insurance Provider'].map(provider_rates)

# Missing coverage = provider share x consultation price
df_imputed['Insurance Coverage'] = df_imputed['Insurance Coverage'].fillna(expected_rate * df_imputed['Consultation Price'])

# Income changes between visits of the same patient, so we use the patient's median
income_median = df_imputed.groupby('Patient ID')['Approximate Annual Income'].transform('median')
df_imputed['Approximate Annual Income'] = df_imputed['Approximate Annual Income'].fillna(income_median)

# Family history cannot be worked out from other columns, so it is marked as Unknown
df_imputed['Family History'] = df_imputed['Family History'].fillna('Unknown')

# Check: every column should show 0 missing values
print("Missing values after imputation:")
print(df_imputed.isnull().sum())
```

To handle the missing values we used imputation, based on the patient's other visits and on rules we found in the data. Group-mean imputation was rejected because it would hide real differences between patients.

* **Age and Education Level:**
  These do not change for a patient during the six months of data, so a missing value is copied from another visit of the same patient (`ffill().bfill()` grouped by `Patient ID`). This restores the real value instead of inventing a new one.

* **Insurance Provider:**
  A missing provider is first filled with the provider of the same patient's other visits. For patients with no provider in any visit, we infer it from the share of the price that the insurance paid (`Insurance Coverage / Consultation Price`). Each provider always pays the same share (None 0%, C 70%, A 80%, B 90%, D 100%), so this share identifies the provider. This rule holds in the data: the median ratio of each provider equals its rate, and only 5 of the 8,015 complete visits break it.

* **Insurance Coverage:**
  A missing coverage is calculated as the provider's share times the consultation price. Since this rule holds in almost all the complete rows, the result is reliable. None of the visits with missing coverage has a wrong price (above £1,000), so the result is not affected by the price errors corrected in 2.5.

* **Approximate Annual Income:**
  Income changes slightly from visit to visit, so a missing value is replaced by the median income of the same patient (`transform('median')`). The median is not affected by a single unusual value, every patient has at least one value, and patients with zero income (children and students) keep their zero. The same median is used as the income feature in the ABT (3.1.1).

* **Family History:**
  This cannot be inferred from any other column, so a missing value is labelled `'Unknown'` instead of being guessed. The value `'None'` is not missing: it means the patient has no family disease and is kept as a category.

20 visits are still missing both the provider and the coverage, because the patient has no provider in any visit and the coverage of these visits is also missing, so there is no information to fill them.

  <b>2.4.3. Inconsistent Age, Profession and Education</b>

We made two rules when relating Age to Profession and Education Level.

#### Rule A: Minors (< 18) who are not marked as Student must have their Profession aligned to 'Student'.

First we will check if any patient older than 2 years old is being considered a Infant.

```python
df_inconsistent.loc[(df_inconsistent['Age']> 2) & (df_inconsistent["Profession"] == "Infant")]
```

Since there are no infants with more than 2 years old, Every minor with more than 2 years should be a student.
First we will check if we have minors with profession different from student or infant.

```python
df_inconsistent.loc[((df_inconsistent["Age"] < 18) & (df_inconsistent["Profession"] != "Student")) & ((df_inconsistent["Age"] < 18) & (df_inconsistent["Profession"] != "Infant"))]
```

There are 108 rows where we the profession should be student or infant but instead is another one.
We will now fix this.

```python
df_inconsistent.loc[
    ((df_inconsistent["Age"] < 18) & (df_inconsistent["Profession"] != "Student")) & ((df_inconsistent['Age'] < 18) & (df_inconsistent['Profession'] != 'Infant')) ,
    "Profession",
] = "Student"
```

#### Rule B: Toddlers (<= 2 years old) must be classified as 'Infant' for both Profession and Education Level.

```python
df_inconsistent.loc[df_inconsistent["Age"] <= 2, ["Profession", "Education Level"]] = "Infant"
```

```python
outlier_summary = []

for var in num_vars:
    series = df_inconsistent[var].dropna()
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    
    outliers = series[(series < lower_bound) | (series > upper_bound)]
    
    outlier_summary.append({
        'Variable': var,
        'Q1 (25%)': round(q1, 2),
        'Q3 (75%)': round(q3, 2),
        'IQR': round(iqr, 2),
        'Lower Bound': round(lower_bound, 2),
        'Upper Bound': round(upper_bound, 2),
        'Outliers Count': len(outliers),
        '% Outliers': round((len(outliers) / len(series)) * 100, 2)
    })

df_outliers = pd.DataFrame(outlier_summary)
display(df_outliers)
```

#### **Logical Inconsistencies & Data Entry Errors**
* **Age:** Ages above 110 (118 to 195) have an extra "1" and are corrected by subtracting 100.
* **Approximate Annual Income & Consultation Price:** Multi-million incomes (£11M–£12M) and consultation prices in the tens of thousands (£25k–£40k) have extra zeros and are corrected by division.
* **Consultation Duration:** Durations between 228 and 600 minutes are not plausible for a consultation and are imputed with the median duration of the department.
* **Insurance Coverage:** Coverage higher than the consultation price is not possible and is recomputed as provider rate x price.

#### **Domain-Plausible Extreme Values**
* **Consultation Price & Insurance Coverage:** The other values beyond the upper whiskers ($1.5 \times \text{IQR}$) are plausible, because the price depends on the department and the coverage on the provider, so they are kept to preserve genuine patient profile variability.

<div style="background-color:#297681; padding:20px; border-radius:10px; border-left:8px"> 
    
**To address each outlier, the following decisions were made:**

| Variable | Outlier | Decision | Why |
|---|---|---|---|
| Age | 82 rows (4 patients) above 110 | Correct: subtract 100 | Typing error with an extra "1", confirmed by patient 1198 (corrected in 2.4) and by the profession and education |
| Consultation Price | 7 rows above £1000 | Correct: divide by 100 | The result is inside the normal range of the department |
| Approximate Annual Income | 21 rows above £1M | Correct: divide by 10 (£1M-£10M) or by 100 (above £10M) | The result is close to the highest normal incomes and to the patient's other visits |
| Income = 0 | 110 patients | Keep | Valid value (mostly children and students) |
| Consultation Duration | 10 rows above 120 min | Impute with the median of the department | The extreme values are not a clear multiple of a normal duration, so the real value is unknown |
| Insurance Coverage | 6 rows higher than the price | Recompute: provider rate x price | All the other rows follow this rule exactly |

We do not remove any row or patient: the errors have a clear pattern or can be replaced, and removing visits would lose real information (and patients).

```python
df_cleaned = df_inconsistent.copy()

# 1. Age: > 110 anos têm um "1" extra -> subtrair 100
df_cleaned.loc[df_cleaned["Age"] > 110, "Age"] = df_cleaned["Age"] - 100

# 2. Consultation Price: > £1000 -> dividir por 100
df_cleaned.loc[df_cleaned["Consultation Price"] > 1000, "Consultation Price"] = (
    df_cleaned["Consultation Price"] / 100
)

# 3. Approximate Annual Income: ajustar consoante o escalão (> 10M div por 100; 1M a 10M div por 10)
income = df_cleaned["Approximate Annual Income"]
df_cleaned.loc[income > 10_000_000, "Approximate Annual Income"] = income / 100
df_cleaned.loc[(income >= 1_000_000) & (income <= 10_000_000), "Approximate Annual Income"] = income / 10

# 4. Consultation Duration: > 120 min -> imputar com a mediana do departamento
dept_median_duration = df_cleaned.groupby("Department")["Consultation Duration"].transform("median")
df_cleaned.loc[df_cleaned["Consultation Duration"] > 120, "Consultation Duration"] = dept_median_duration

# 5. Insurance Coverage: se cobertura > preço -> recalcular (provider rate * price)
provider_rates = {'None': 0.0, 'Provider A': 0.8, 'Provider B': 0.9, 'Provider C': 0.7, 'Provider D': 1.0}
expected_coverage = (df_cleaned["Consultation Price"] * df_cleaned["Insurance Provider"].map(provider_rates).astype(float)).round(2)
invalid_coverage = df_cleaned["Insurance Coverage"] > df_cleaned["Consultation Price"]
df_cleaned.loc[invalid_coverage, "Insurance Coverage"] = expected_coverage[invalid_coverage]

# Verificação dos sumários estatísticos pós-limpeza
display(df_cleaned[num_cols].describe().round(2))
```

```python
static_attributes = df_cleaned.groupby("Patient ID").agg({
    "Age": "first",
    "Gender": "first",
    "City of Residence": "first",
    "Profession": "first",
    "Education Level": "first",
    "Marital Status": "first",
    "Family History": "first",
    "Approximate Annual Income": "median",
    "Insurance Provider": "first"
}).reset_index()
static_attributes.columns = ['Patient_ID', 'Age', 'Gender', 'City', 'Profession', 'Education_Level', 'Marital_Status', 'Family_History', 'Annual_Income', 'Insurance_Provider']

static_attributes.head()
```

  <b>3.1.2 Visit Behaviour Features </b>
</div>


These features describe how each patient uses the hospital: how many consultations they had, in how many different departments, which department they visit most and how long their consultations last.

```python
visits = df_cleaned.groupby('Patient ID').agg({
    'Visit Date': 'count',
    'Department': ['nunique', lambda s: s.mode()[0]],
    'Consultation Duration': ['mean', 'sum']
}).reset_index()
visits.columns = ['Patient_ID', 'N_Visits', 'N_Departments', 'Most_Visited_Department', 'Avg_Duration', 'Total_Duration']
visits.round(3).head()
```

<a id="sec-3-1-3"></a>
<div style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #355473; color:#355473;">
  <b>3.1.3 Financial Features </b>
</div>


The financial features show how much each patient's consultations cost, how much the insurance paid and how much the patient paid (out-of-pocket = price - coverage). The `coverage_share` is the part of the total cost paid by the insurance.

```python
# Out-of-pocket amount of each visit (price - coverage)
df_cleaned['Out of Pocket'] = df_cleaned['Consultation Price'] - df_cleaned['Insurance Coverage']

financial = df_cleaned.groupby('Patient ID').agg({
    'Consultation Price': ['sum', 'mean'],
    'Insurance Coverage': 'sum',
    'Out of Pocket': ['sum', 'mean']
}).reset_index()
financial.columns = ['Patient_ID', 'Total_Price', 'Avg_Price', 'Total_Coverage', 'Total_Out_Of_Pocket', 'Avg_Out_Of_Pocket']

financial['Coverage_Share'] = financial['Total_Coverage'] / financial['Total_Price']
financial.round(3).head()
```

<a id="sec-3-1-4"></a>
<div style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #355473; color:#355473;">
  <b>3.1.4 Satisfaction Features </b>


The satisfaction features use only the valid ratings (1-5); the missing ones are ignored by `mean` and `min`. A low rating is a rating of 1 or 2. We also keep the number of invalid ratings of each patient as a data quality flag.

```python
satisfaction = df_cleaned.groupby('Patient ID').agg({
    'Satisfaction Level': ['mean', 'median', 'min', 'max'],
    'Low Rating': 'mean',
    'Invalid Rating': 'sum'
}).reset_index()
satisfaction.columns = ['Patient_ID', 'Avg_Satisfaction', 'Median_Satisfaction', 'Min_Satisfaction', 'Max_Satisfaction', 'Share_Low_Ratings', 'N_Invalid_Ratings']
satisfaction.round(3).head()
```

<a id="sec-3-1-5"></a>
<div style="background-color:#EFF5FF; padding:20px; border-radius:10px; border-left:8px solid #355473; color:#355473;">
  <b>3.1.5 Recency and Activity Features </b>


Recency is measured in days between the first and the last visit and the reference date (1 July 2024). Activity is measured by the number of different months with at least one visit and by the average number of visits per month in the six months of data.

```python
recency = df_cleaned.groupby('Patient ID').agg({
    'Visit Date': ['min', 'max', lambda s: s.dt.month.nunique()]
}).reset_index()
recency.columns = ['Patient_ID', 'First_Visit', 'Last_Visit', 'N_Active_Months']

# Number of visits of each patient
n_visits = recency['Patient_ID'].map(df_cleaned.groupby('Patient ID').size())

# Recency: days between the first/last visit and the reference date (1 July 2024)
recency['Recency_Days'] = (ref_date - recency['Last_Visit']).dt.days
recency['Days_Since_First_Visit'] = (ref_date - recency['First_Visit']).dt.days

# Tenure: days between the first and the last visit
recency['Tenure_Days'] = (recency['Last_Visit'] - recency['First_Visit']).dt.days

# Average interval between consecutive visits
recency['Avg_Days_Between_Visits'] = (recency['Tenure_Days'] / (n_visits - 1)).round(2)

# Activity: visits per month in the six months of data
recency['Visits_Per_Month'] = (n_visits / 6).round(2)
recency.head()
```

```python
abt = static_attributes.merge(visits, on='Patient_ID').merge(financial, on='Patient_ID').merge(satisfaction, on='Patient_ID').merge(recency, on='Patient_ID')
abt = abt.round(2)

print("ABT shape:", abt.shape)
print("One row per patient:", abt['Patient_ID'].is_unique)
print("Missing values:", abt.isna().sum().sum())
```

Together with the previous charts, this shows that satisfaction does not depend on who the patient is or how they pay. Low ratings are spread across all segments, so improvements should focus on the consultation experience itself, especially in General Practice.

To understand whether patients rate their visits consistently, we calculate the standard deviation of each patient's ratings: a low value means similar ratings in every visit, while a high value means the same patient rates their visits very differently.

```python
# Standard deviation of the ratings of each patient
ABT_city_hospital['Satisfaction_Volatility'] = ABT_city_hospital['Patient_ID'].map(df_cleaned.groupby('Patient ID')['Satisfaction Level'].std())

plt.figure(figsize=(9, 5))
sns.scatterplot(data=ABT_city_hospital, x='N_Visits', y='Satisfaction_Volatility', hue='Avg_Satisfaction', palette=DIV, alpha=0.7)
plt.title('Satisfaction consistency vs. number of visits')
plt.xlabel('Number of visits')
plt.ylabel('Standard deviation of the ratings')
plt.legend(title='Avg satisfaction', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.show()
```

Each point is a patient, placed by the number of visits (x-axis) and by the standard deviation of their ratings (y-axis), and coloured by their average satisfaction. Most patients have a high standard deviation, which means that the same patient often gives very different ratings from one visit to another.

The points form a cloud with no clear pattern: patients with few and with many visits have similar variability, and patients with low (coral) and high (teal) average satisfaction appear at every level of variability. This suggests that satisfaction depends more on each consultation than on the patient.
